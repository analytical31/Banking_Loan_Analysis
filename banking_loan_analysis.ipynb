Banking Loan Data Analysis - using Python

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

It's Time to Load Dataset

```python
df=pd.read_csv("Bank_Loan_Dataset.csv")
print(df.head())
```

Lets see Visual Analysis

```python
# 1. visul- Loan Status
plt.Figure(figsize=(6,2))
sns.countplot(x='Loan_Status', data=df, color='yellow')
plt.title("Status of Loan")
plt.show()
```

```python
#2. visual- Loan Type
plt.Figure(figsize=(4,2))
sns.countplot(x='Loan_Type', data=df, color='orange')
plt.title("Types of Loan")
plt.show()
```

```python
# 3. visual- Income Distribution

plt.hist(df['Annual_Income'], bins=10, color='purple')
plt.title("Income Distribution")

plt.show()
```

```python
# 4. visual- Credit Score Distribution

plt.hist(df['Credit_Score'], bins=10)
plt.title("Credit Score Distribution")
plt.show()
```

```python
# 5. visual- Loan Amount by Loan Type

sns.boxplot(
    x='Loan_Type',
    y='Loan_Amount',
    data=df, color='red'
)
plt.title("Loan Amount by Loan Type")
plt.show()
```

```python
# 6. visual- Age Distribution

plt.hist(df['Age'], color='orange')
plt.title("Age Distribrution")
plt.show()
```

```python
# 7. visual-Employment Status

sns.countplot(
    x='Employment_Status', color='green',
    data=df 
)
plt.title("Employment Status")
plt.show()
```

```python
# 8. visual-Correlation Heatmap

sns.heatmap(
    df[['Age',
        'Annual_Income',
        'Loan_Amount',
        'Credit_Score']].corr(),
        annot=True
)
plt.title("Correlation")
plt.show()
```
